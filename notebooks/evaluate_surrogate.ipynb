# Option-2 NN Surrogate Evaluation (Issue #4075)

Evaluate the trained FiLM-MLP surrogate model on test set.


```python
import numpy as np
import torch
from pathlib import Path

from src.shared.python.motion_matching.dataset import load_sweep_dataset, make_synthetic_sweep
from src.shared.python.motion_matching.surrogate import train_surrogate, TrainConfig
```

## Load Dataset and Train Model

```python
# Use synthetic dataset for testing
dataset_path = Path("data/sweep_synthetic_eval")
dataset_path.mkdir(parents=True, exist_ok=True)
make_synthetic_sweep(dataset_path, n_trials=100, n_joints=14, seed=42)

dataset = load_sweep_dataset(dataset_path, lazy=False)
print(f"Loaded {dataset.n_trials()} trials with {dataset.n_joints()} joints")
```

```python
# Train the model
train_cfg = TrainConfig(
    n_epochs=10,
    batch_size=16,
    device="cpu",
)

print("Training surrogate model...")
trained = train_surrogate(dataset, train_cfg)
print(f"Final validation loss: {trained.curves.val_loss[-1]:.6f}")
```

## Model Architecture Summary

```python
cfg = trained.config
total_params = sum(p.numel() for p in trained.model.parameters())

print(f"Model parameters: {total_params:,}")
print(f"coeff_dim (D): {cfg.coeff_dim}")
print(f"seq_len (N): {cfg.seq_len}")
print(f"hidden_dim: {cfg.hidden_dim}")
print(f"n_layers: {cfg.n_layers}")
```
